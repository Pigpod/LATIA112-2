```python
import pandas as pd

#csv_file = './111_student.csv'
csv_file ="./111_student.csv"

df = pd.read_csv(csv_file) # 讀取csv檔案
print(df.head())

#查看資料欄位資訊
print(df.info())
```

# 題目一 : 哪間大專院校有最多的博士生


```python
df1 = df[['學校名稱','等級別','總計'] ] #從df取出［縣市名稱、等級別、總計]欄位放入新的df1中

df1_1 = df[df1['等級別'] == 'D 博士'] #從df1中篩選出[D 博士]]的資料放入df1_1中
df1_1_sorted = df1_1.sort_values(by='總計',ascending=False) #利用.sort_values 依[總計]欄位降冪排列，將排列好的資料存進df1_1_sorted
print("112學年度在籍的博士生最多人數之學校為", df1_1_sorted.head(1)['學校名稱'].values) #利用.head(1)將第一筆資料中的[學校名稱].values顯示出來
print("共",df1_1.head(1)["總計"].values,"位博士生") #利用.head(1)將第一筆資料中的[總計].values顯示出來
```

# 題目二 : 位於新北市的大專院校總共有幾間


```python
df2 = df[['縣市名稱']]  #從df取出［縣市名稱]欄位放入新的df2中

df2_1 = df[df2['縣市名稱'] == '01 新北市'] #從df2中篩選出[01 新北市]的資料放入df2_1中
df2_2 = df2_1.drop_duplicates('學校代碼') #將df2_1中具有同樣學校代碼的資料拋棄，並存入df2_2

print("新北市共",len(df2_2),"間大專院校") #利用len求出df2_2中總共有幾筆資料，相當於新北市總共有幾間大專院校
```

# 題目三 : 哪間大專院校擁有最多學生

```python


count = {}      #創建一個空字典 用於儲存每間學校的總人數

def create_count_dict(df,count):    #建立函式將df3中"學校名稱"的各個學校名稱寫入count，每個學校的資料(總人數)均設為0
  df3 = df.drop_duplicates('學校代碼') #將df中具有同樣學校代碼的資料拋棄，並存入df3
  for i in df3['學校名稱']:
    count[i] = 0

def count_every_schools_data(df):   #建立函式
  for j in range(len(df)):    #使用 range(len(df)) 拜訪 DataFrame df 中的每一行
      count[df.iloc[j,2]] += int(df.iloc[j,5])  #count[df.iloc[j,2]] 取得該學校的字典，int(df.iloc[j,5]) 將df[j,5]也就是"總計"欄的數據寫入對應的學校中

create_count_dict(df,count)     #執行函式
count_every_schools_data(df)    #執行函式


df3_1 = df.drop_duplicates('學校代碼')  #將df中具有同樣學校代碼的資料拋棄，並存入df3_1


df3_1['總人數'] = count.values()    #在df3_1中新增總人數的空欄位，並將count字典中紀錄的數字依序填入

df3_2 = df3_1.sort_values(by='總人數',ascending=False)  #將df3_1依照總人數由大到小排列並存成df3_2
print("111學年度在校生(含延修生)最多的學校為:", df3_2.head(1)['學校名稱'].values, '人數為' ,df3_2.head(1)['總人數'].values) #列印出df3_2的head(1)，也就是總人數最多的學校名稱以及總人數為何





```
